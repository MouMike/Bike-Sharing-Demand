```
import pandas as pd
import matplotlib.pyplot as plt
import matplotlib.dates as mdates
import seaborn as sns
from autogluon.tabular import TabularPredictor
import gc
```

```
# Create the train dataset in pandas by reading the csv
# Set the parsing of the datetime column so you can use some of the `dt` features in pandas later
train = pd.read_csv("train.csv")
train["datetime"] = pd.to_datetime(train["datetime"])
train.head()
```

```
# Create the test pandas dataframe in pandas by reading the csv, remember to parse the datetime!
test = pd.read_csv("test.csv")
test["datetime"] = pd.to_datetime(test["datetime"])
test.head()
```

```
# Simple output of the train dataset to view some of the min/max/variation of the dataset features.
print(f"Number of months in training dataset: {train['datetime'].dt.to_period('M').nunique()}")
print(f"Number of months in test dataset: {test['datetime'].dt.to_period('M').nunique()}")

last_training_date = train["datetime"].max()
first_training_date = train["datetime"].min()
first_test_date = test["datetime"].min()
last_test_date = test["datetime"].max()

print(f"\nLast ride in training set: {last_training_date}")
print(f"First ride in test set: {first_test_date}")

if((first_training_date < first_test_date) & (last_training_date < first_test_date) 
  | (first_test_date < first_training_date) & (last_test_date < first_training_date)):
    print("\nTraining dataset and test dataset chosen sequentially.")
else:
    print("\nTraining dataset and test dataset chosen randomly.")
    
print(f"\nAverage number of rides per hour: {train["count"].mean()}")
print(f"Count variation: {train["count"].var()}")
```

## Add More Features

```
# create more features

# define train rush hour feature
train["rush_hour"] = train["hour"]
train["rush_hour"] = "other"
train.loc[((train["hour"] >= 7) & (train["hour"] <= 9) & (train["workingday"] == 1)), "rush_hour"] = "morning"
train.loc[((train["hour"] >= 11) & (train["hour"] <= 13) & (train["workingday"] == 1)), "rush_hour"] = "lunch"
train.loc[((train["hour"] >= 16) & (train["hour"] <= 20) & (train["workingday"] == 1)), "rush_hour"] = "evening"
train.loc[((train["hour"] >= 10) & (train["hour"] <= 19) & (train["workingday"] == 0)), "rush_hour"] = "day-off"
#train = train.drop(["hour"], axis=1)

# define test rush hour feature
test["rush_hour"] = test["hour"]
test["rush_hour"] = "other"
test.loc[((test["hour"] >= 7) & (test["hour"] <= 9) & (test["workingday"] == 1)), "rush_hour"] = "morning"
test.loc[((test["hour"] >= 11) & (test["hour"] <= 13) & (test["workingday"] == 1)), "rush_hour"] = "lunch"
test.loc[((test["hour"] >= 16) & (test["hour"] <= 20) & (test["workingday"] == 1)), "rush_hour"] = "evening"
test.loc[((test["hour"] >= 10) & (test["hour"] <= 19) & (test["workingday"] == 0)), "rush_hour"] = "day-off"
#test = test.drop(["hour"], axis=1)

#change data type to category
train["rush_hour"] = train["rush_hour"].astype("category")
test["rush_hour"] = test["rush_hour"].astype("category")
```

```
# define temperature classification feature for training set
train.loc[train["temp"] < 10, "temp_class"] = "cold"
train.loc[((train["temp"] >= 10) & (train["temp"] <= 25)), "temp_class"] = "mild"
train.loc[train["temp"] > 25, "temp_class"] = "hot"
#train = train.drop(["temp"], axis=1)

# define temperature classification feature for test set
test.loc[test["temp"] < 10, "temp_class"] = "cold"
test.loc[((test["temp"] >= 10) & (test["temp"] <= 25)), "temp_class"] = "mild"
test.loc[test["temp"] > 25, "temp_class"] = "hot"
#test = test.drop(["temp"], axis=1)

#change data type to category
train["temp_class"] = train["temp_class"].astype("category")
test["temp_class"] = test["temp_class"].astype("category")
```

```
# define wind classification feature for training set
train.loc[train["windspeed"] <= 25, "wind_class"] = "mild"
train.loc[train["windspeed"] > 25, "wind_class"] = "very-windy"
#train = train.drop(["windspeed"], axis=1)

# define wind classification feature for test set
test.loc[test["windspeed"] <= 25, "wind_class"] = "mild"
test.loc[test["windspeed"] > 25, "wind_class"] = "very-windy"
#test = test.drop(["windspeed"], axis=1)

#change data type to category
train["wind_class"] = train["wind_class"].astype("category")
test["wind_class"] = test["wind_class"].astype("category")
```

```
# define humidity classification feature for training set
train.loc[(train["humidity"] <= 50), "humid_class"] = "not-humid"
train.loc[(train["humidity"] > 50), "humid_class"] = "humid"
#train = train.drop(["humidity"], axis=1)

# define humidity classification feature for test set
test.loc[(test["humidity"] <= 50), "humid_class"] = "not-humid"
test.loc[(test["humidity"] > 50), "humid_class"] = "humid"
#test = test.drop(["humidity"], axis=1)

#change data type to category
train["humid_class"] = train["humid_class"].astype("category")
test["humid_class"] = test["humid_class"].astype("category")
```
